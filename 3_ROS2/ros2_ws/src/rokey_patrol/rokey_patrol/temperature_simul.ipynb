```python
import random
import matplotlib.pyplot as plt
```

```python
sensor_data=[0]
threshold=0.1 #화재 발생 확률
for i in range(100):
    if random.random()<threshold:
        sensor_data.append(25+random.uniform(50, 60))
    else:
        sensor_data.append(25+random.uniform(-1.0, 1.0))


    
```

```python
plt.plot(sensor_data)
```
