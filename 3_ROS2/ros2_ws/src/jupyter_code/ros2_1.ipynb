```python
import rclpy
from std_msgs.msg import String

# 이미 초기화되어 있지 않은 경우에만 init() 호출
if not rclpy.ok():
    rclpy.init()

node = rclpy.create_node("nb_node")
pub = node.create_publisher(String, "nb", 10)
```

```python
msg=String()
msg.data="Hello ROS"
pub.publish(msg)
rclpy.spin_once(rclpy.spin_once(node, timeout_sec=1.0))
#test
```

```python
node.destroy_node()
rclpy.shutdown()
```
