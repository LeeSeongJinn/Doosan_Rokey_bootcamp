```python
# 모든 요소 정규분포를 따르는변수와 3개의 행렬
np.random.randn(2,3,4)
# random: 무작위, 임의로(임의적) >> 일반화
# randn (n: normalize 정규화) >> 정규분포 (평균 0, 표준편차 1)
# 표준화 (x - x_bar) / sigma
# >> x :원래 값, x_bar: 평균, sigma 편차(평균에서 떨어진 값)
# 정규화 (x-x_min) / (x_max - x_min) >> [0,1]
# >> x: 원래 값, x_min : 해당 속성(열) 최소값, x_max : 해당 속성(열) 최대값
# n으로 나눈다 n = len(x)
```

응용예시

```python
# 두 변수를 비교해서 일치하는 개수로 정확도 계산

# 2개의 넘파이 배열 준비

# 정답
yt = np.array([1,1,0,1,0,1,1,0,1,0])

# 예측
yp = np.array([1,1,0,1,0,1,0,1,1,1])
```

```python
# 정답(yt)와 예측(yp) 잘 맞았나? (비교)

(yt == yp)
```

```python
matched = (yt == yp)
matched.sum()
```

```python
# 맞은 개수(정답 수 세는 방법)
correct = matched.sum()

# 전체 수
total = len(matched)

# 정확도 계산
accuracy = correct / total
print(accuracy)
```

```python
print(f'정답수: {correct}, 전체 개수: {total}, 정확도:{accuracy*100}%')
```
